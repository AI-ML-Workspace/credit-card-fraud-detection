# Final Model Preparation

## Objective

The objective of this notebook is to prepare the optimized machine learning model for production use by loading the tuned model, validating that it works correctly, making sample predictions, and saving it as the project's final production model.

```python
import os
import pandas as pd
import joblib
```

```python
df = pd.read_csv("../data/processed/processed_creditcard.csv")

df.head()
```

### Observation

The processed credit card fraud detection dataset was successfully loaded from `../data/processed/processed_creditcard.csv`.

```python
X = df.drop("Class", axis=1)

Y = df["Class"]
```

### Feature & Target Separation

Only the feature matrix (`X`) is required to generate model predictions, while the target column (`Y`) is retained solely for ground truth comparison.

```python
best_model = joblib.load("../models/random_forest_tuned.pkl")

print("Final tuned model loaded successfully.")
```

```python
best_model
```

### Observation

The tuned Random Forest model has been loaded successfully and is ready for inference.

```python
sample_data = X.head()

sample_data
```

```python
predictions = best_model.predict(sample_data)

print(predictions)
```

```python
print(Y.head())
```

### Observation

The predicted class labels match the actual ground truth labels for the sample rows, demonstrating that the model successfully performs inference on input features.

```python
joblib.dump(best_model, "../models/final_model.pkl")

print("Final production model saved successfully.")
```
