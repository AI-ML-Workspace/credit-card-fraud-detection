#  Data Analysis (DA)

## Objective

The objective of this notebook is to explore the dataset visually and statistically to identify patterns, trends, anomalies, and relationships before preprocessing and model training.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("../data/raw/creditcard.csv")
```

## Distribution of transcation amount 

```python
df["Amount"].describe()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Amount"], bins=50)
plt.title("Transaction Amount Distribution")
plt.xlabel("Amount")
plt.ylabel("Frequency")
plt.show()
```

## Distribution of time 

```python
plt.figure(figsize=(8,5))
plt.hist(df["Time"], bins=50)
plt.title("Transaction Time Distribution")
plt.xlabel("Time")
plt.ylabel("Frequency")
plt.show()
```

## Fraud vs Legit

```python
class_counts = df["Class"].value_counts()
class_counts.plot(kind="bar")
plt.title("Fraud vs legitimate tansactions")
plt.xlabel("Class")
plt.ylabel("Number of Transactions")
plt.xticks([0, 1], ["Legitimate (0)", "Fraud (1)"], rotation=0)

plt.show()
```

```python
import seaborn as sns

plt.figure(figsize=(18,12))
sns.heatmap(df.corr(), cmap="coolwarm")
plt.title("Correlation Matrix")
plt.show()
```

```python
corr = df.corr()["Class"].sort_values()

plt.figure(figsize=(8,10))
corr.plot(kind="barh")
plt.title("Correlation with Class")
plt.show()
```

variables have positive or negative correaltion with fraud
